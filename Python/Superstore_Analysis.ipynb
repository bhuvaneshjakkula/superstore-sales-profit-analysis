```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales
```

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
category_profit
```

```python
category_avg_discount = df.groupby('Category')['Discount'].mean().sort_values(ascending=False)
category_avg_discount
```

```python
category_summary = df.groupby('Category').agg({'Sales': 'sum','Profit': 'sum'})

category_summary['Profit Margin (%)'] = (category_summary['Profit']/ category_summary['Sales']) * 100

category_summary.sort_values('Profit Margin (%)',ascending=False)
```

```python
discount_profit = df.groupby('Discount')['Profit'].mean().sort_index()
discount_profit
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
region_sales
```

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
region_profit
```

```python
region_summary = df.groupby('Region').agg({'Sales':'sum', 'Profit':'sum'})

region_summary['Profit Margin(%)'] = (region_summary['Profit'] / region_summary['Sales']) * 100

region_summary.sort_values('Profit Margin(%)', ascending=False)
```

```python
state_sales = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)
state_sales
```

```python
state_profit = df.groupby('State')['Profit'].sum().sort_values(ascending=False).head(10)
state_profit
```

```python
state_summary = (df.groupby('State').agg({'Sales': 'sum','Profit': 'sum'}))

state_summary['Profit Margin (%)'] = (state_summary['Profit'] /state_summary['Sales']) * 100

state_summary.sort_values('Sales',ascending=False)
```

```python
customers_generate_loss = df.groupby('Customer ID')['Profit'].sum().sort_values()
```

```python
loss_customers = customers_generate_loss[customers_generate_loss < 0]
print(loss_customers.head(10))
```

```python
customer_summary = df.groupby("Customer ID").agg(total_sales=("Sales", "sum"),total_profit=("Profit", "sum")).reset_index()

top_sales = customer_summary.sort_values("total_sales", ascending=False).head(10)
top_profit = customer_summary.sort_values("total_profit", ascending=False).head(10)

overlap = set(top_sales["Customer ID"]) & set(top_profit["Customer ID"])
print(f"\n{len(overlap)} of the top 10 sales customers are also top 10 profit customers:")
print(overlap)
```
